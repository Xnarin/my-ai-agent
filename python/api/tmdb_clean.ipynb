```python
import requests
from pprint import pprint
import os
from dotenv import load_dotenv

load_dotenv() # .env에 TMDB_TOKEN 넣어두기 
```

```python
def get_movies(what="now_playing", page=1):
    # movie/ 뒤에 now_playing or popular
    url = f"https://api.themoviedb.org/3/movie/{what}"

    token = os.getenv("TMDB_TOKEN")

    my_params = {
        "language" : "ko-kr",
        "page" : page
        # region
    }

    my_headers = {
        "Authorization" : f"Bearer {token}",
    }

    response = requests.get(url, headers=my_headers, params=my_params)

    # 함수 내부에서는 raise만 하고, except 처리는 밖에서 진행합니다.
    # 항상 그래야하는건 아닙니다.
    response.raise_for_status()

    data = response.json()
    
    results = data.get("results")
    return results

def get_best_movie(movies):
    
    best_movie = max(movies, key=lambda movie : movie['vote_average'])

    return best_movie
```

### 문제 1
현재 상영 중인 영화(Now Playing) 중 평점(vote_average)이 가장 높은 영화를 찾으시오.

```python

try:
    # 1. now_playing 영화 데이터 가져올꺼야.
    movies = get_movies()

    # 2. 평점 높은 영화 가져올꺼야.
    best_movie = get_best_movie(movies)
    print(best_movie['title'], best_movie['vote_average'])
except Exception as e:
    print('오류 발생!')
    print(e)    

```

### 문제 3
현재 상영 중인 영화(now_playing) 데이터에서 다음 정보만 담긴 리스트를 만드세요.

- title (영화 제목)  
- vote_average (평점)  
- release_date (개봉일)

```python
# movie -> simple movie를 함수로 바꿔보자.
def get_simple_movie(movie):
    title = movie['title']
    vote_average = movie['vote_average']
    release_date = movie['release_date']

    simple_movie = {
        "title" : title,
        "vote_average" : vote_average,
        "release_date" : release_date,
    }
    return simple_movie

# 추가문제 : 위 함수를 변경해서 내가 원하는 key들로 simple_movie를 정하게 하는 것을 만들어 보자.

try:
    movies = get_movies()

    simple_movies = [get_simple_movie(movie) for movie in movies]

    pprint(simple_movies)
except Exception as e:
    print(e)
```

### 문제 4
현재 상영 중인 영화(now_playing) 데이터를 vote_average(평점)순으로 정렬하여 영화 제목이 담긴 리스트를 만드세요.
단, 리스트의 길이는 40 이상이어야 합니다.


```python
try:
    # 아래 로직을 함수로 만들어서
    # n페이지까지의 정보를 한번에 가져오는 함수를 만들수도 있을겁니다.
    pages = [1, 2, 3, 4]
    many_movies = []
    for page in pages:

        movies = get_movies(page=page)
        # many_movies += movies
        many_movies.extend(movies)
    # pprint([movie['title'] for movie in many_movies])
    # many_movies를 활용해서 vote_aveate 순으로 정렬 -> 영화 제목만 가져오기

    # sorted 
    # key 사용해서 기준 잡기
    # reverse=True -> 역순으로 정렬

    sorted_movies = sorted(many_movies, key=lambda movie : movie['vote_average'], reverse=True)

    sorted_movies_title = [(movie['title'], movie['vote_average']) for movie in sorted_movies]

    pprint(sorted_movies_title)


except Exception as e:
    print(e)
```

### 문제 5
현재 상영 중인 영화 중 평점(vote_average)이 가장 높은 영화의 수익(revenue)를 찾으시오.  
hint) API 문서의 endpoint 항목을 참고하세요.

```python
def get_movie_by_id(movie_id):
    url = f"https://api.themoviedb.org/3/movie/{movie_id}"

    token = os.getenv("TMDB_TOKEN")

    my_params = {
        # language
        # region
    }

    my_headers = {
        "Authorization" : f"Bearer {token}",
    }

    response = requests.get(url, headers=my_headers, params=my_params)

    # 함수 내부에서는 raise만 하고, except 처리는 밖에서 진행합니다.
    # 항상 그래야하는건 아닙니다.
    response.raise_for_status()

    data = response.json()
    
    return data
```

```python
try:

    # 1. 현재 상영중인 영화 가져오기
    movies = get_movies()
    # 2. 평점이 가장 높은 영화 가져오기
    movie = get_best_movie(movies)

    # 2.a revenue가 0이라 첫번째 영화로 대체하기
    movie = movies[0]
    # print(movie)

    movie_id = movie.get('id')
    # 3. 그 영화의 detail 정보 가져오기 -> revenue 찾기.
    movie_detail = get_movie_by_id(movie_id)

    # pprint(movie_detail)
    revenue = movie_detail.get('revenue')

    print(revenue)

except Exception as e:
    print(e)
```
