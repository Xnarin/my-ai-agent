    1. `시도별 실시간 측정정보 조회`의 `제주`의 데이터에 대해, 초 미세먼지 농도가 가장 낮은 `stationName`을 찾으시오.

```python
try:
    url = "https://apis.data.go.kr/B552584/ArpltnInforInqireSvc/getCtprvnRltmMesureDnsty?serviceKey=%2BiEaRW7xrORUYANDjKPrnvW9DAAqDJNKv3E4sm3Vwbes8db4rFSa%2FTnEVPEmaCWv1BzeVE2ek9Fv8onYt9obpQ%3D%3D&returnType=json&numOfRows=100&pageNo=1&sidoName=%EC%84%9C%EC%9A%B8&ver=1.0"

    response = requests.get(url)
    response.raise_for_status()

    json_data = response.json()

    response = json_data.get('response')
    body = response.get('body')
    items = body.get('items')

    min_value = float('inf')
    min_item = None
    for item in items:
        pm25Value = item.get('pm25Value')
        
        if pm25Value == '-':
            continue
        if min_value > int(pm25Value):
            min_value = int(pm25Value)
            min_item = item
    print(min_value, min_item.get('stationName'))
except Exception as e:
    print(e)
    print(e.with_traceback())
```

```python
stations = {}

for item in items:
    station_name = item.get("stationName")

    if station_name:
        stations[station_name] = item

pprint(stations)
```
