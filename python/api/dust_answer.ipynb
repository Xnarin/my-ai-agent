```python
# timeout(504)이 발생한다 -> 500번대 -> 서버잘못.

# 1. 이따가 다시 시도한다. 근데 자동으로.

# 2. 이걸 json으로 저장해서 그걸 api인셈 치고 사용한다.
```

```python
# 1. 이따가 다시 시도한다. 근데 자동으로.

# for _ in range(횟수):
#     request 해봐

#     되? 그럼 굿

#     안되? 그럼 기다렸다가 다시 해.

import time

def get_dust_info(attempt = 3):

    for _ in range(attempt):
        try:
            url = "https://apis.data.go.kr/B552584/ArpltnInforInqireSvc/getCtprvnRltmMesureDnsty?serviceKey=%2BiEaRW7xrORUYANDjKPrnvW9DAAqDJNKv3E4sm3Vwbes8db4rFSa%2FTnEVPEmaCWv1BzeVE2ek9Fv8onYt9obpQ%3D%3D"

            params = {
                # "serviceKey" : "%2BiEaRW7xrORUYANDjKPrnvW9DAAqDJNKv3E4sm3Vwbes8db4rFSa%2FTnEVPEmaCWv1BzeVE2ek9Fv8onYt9obpQ%3D%3D",
                "returnType" : 'json',
                'numOfRows' : 100,
                'pageNo' : 1,
                'sidoName' : '서울',
                'ver' : 1.0
            }
            response = requests.get(url, params=params)

            # 504같은 정상적이지 않은 응답들을 raise합니다.
            response.raise_for_status()

            json_data = response.json()

            response = json_data.get('response')
            body = response.get('body')
            items = body.get('items')
            return items
        except Exception as e:
            print('오류발생 ㅠㅠ')
            print(e)
            time.sleep(1)

data = get_dust_info()

print(data)
```
