### 문제 1
현재 상영 중인 영화(Now Playing) 중 평점(vote_average)이 가장 높은 영화를 찾으시오.

```python
import requests
import os
from pprint import pprint
from dotenv import load_dotenv

# .env 파일의 내용을 환경변수로 로드
load_dotenv()

# JWT - API TOKEN과 header을 이용한 인증 방식
token = os.getenv('TMDB_TOKEN')


if not token:
    raise ValueError("TMDB_TOKEN을 찾지 못했습니다. .env 파일 위치와 변수명을 확인하세요.")

url = "https://api.themoviedb.org/3/movie/now_playing"

  

my_headers = {
    "Authorization" : f"Bearer {token}"
} # 정보 보낼 때 플러스 알파인 것

try :
    # response = requests.get(url)
    # headers 매개 변수를 받고 -> header
    response = requests.get(url, headers = my_headers)
    response.raise_for_status()

    movies = response.json()["results"]

    best_movie = max(movies, key=lambda movie: (movie["vote_average"], movie["vote_count"]))


    print("평점이 가장 높은 영화")
    print("제목:", best_movie["title"])
    print("평점:", best_movie["vote_average"])
    print("투표 수:", best_movie["vote_count"])

except Exception as e:
    print("오류발생")
    print(e)
```

### 문제 3
현재 상영 중인 영화(now_playing) 데이터에서 다음 정보만 담긴 리스트를 만드세요.

- title (영화 제목)  
- vote_average (평점)  
- release_date (개봉일)

```python
def get_movies(what="now_playing", page = 1):
    # movie/ 뒤에 now_playing or popular
    url = f"https://api.themoviedb.org/3/movie/{what}"

    token = os.getenv("TMDB_TOKEN")

    my_headers = {
        "Authorization" : f"Bearer {token}",
    }

    params = {
        "language" : "ko-kr",
        "page": page
    }

    response = requests.get(url, headers=my_headers, params=params)

    # 함수 내부에서는 raise만 하고, except 처리는 밖에서 진행합니다.
    # 항상 그래야하는건 아닙니다.
    response.raise_for_status()

    data = response.json()
    
    results = data.get("results")
    return results

def get_best_movie(movies):
    
    best_movie = max(movies, key=lambda movie : movie['vote_average'])

    return best_movie
```

```python
try:
    movies = get_movies()
    simple_movies = []

    for movie in movies:
        # movie에서 정보를 3개만 빼다가 새로운 dict만들고
        title = movie['title']
        vote_average = movie['vote_average']
        release_date = movie['release_date']

        simple_movie = {
            "title" : title,
            "vote_average" : vote_average,
            "release_date" : release_date,
        }
        # print(simple_movie)

        # 그거를 list에 넣을꺼야.
        simple_movies.append(simple_movie)

    pprint(simple_movies)
except Exception as e:
    print(e)
```

### 문제 4
현재 상영 중인 영화(now_playing) 데이터를 vote_average(평점)순으로 정렬하여 영화 제목이 담긴 리스트를 만드세요.
단, 리스트의 길이는 40 이상이어야 합니다.


```python
try :
    # response = requests.get(url)
    # headers 매개 변수를 받고 -> header
    movies_page1 = get_movies("now_playing", 1)
    movies_page2 = get_movies("now_playing", 2)
    movies = movies_page1 + movies_page2

    movie_lst = []

    for i in range(len(movies) - 1) :
        for j in range(len(movies) - 1 - i) :
            left_vote = movies[j].get("vote_average")
            right_vote = movies[j+1].get("vote_average")

            if(left_vote < right_vote) :
                movies[j], movies[j+1] = movies[j+1], movies[j]

    for movie in movies :
        movie_lst.append(movie.get("title"))

    pprint(movie_lst)

except Exception as e:
    print("오류발생")
    print(e)
```

hint) API 문서의 endpoint 항목을 참고하세요.

```python
def get_movie_revenue(movie_id):
    url = f"https://api.themoviedb.org/3/movie/{movie_id}"

    token = os.getenv("TMDB_TOKEN")

    headers = {
        "Authorization": f"Bearer {token}"
    }

    response = requests.get(url, headers=headers)
    response.raise_for_status()

    movie_detail = response.json()

    return movie_detail.get("revenue")
```

### 문제 8
현재 상영 중인 영화(now_playing) 데이터에서 다음 정보만 담긴 리스트를 만드세요.
단, 장르는 text로 이루어진 list입니다.

- title (영화 제목)  
- genres (장르)

hint) API 문서의 endpoint 항목을 참고하세요


```python
try:
    movies = get_movies()

    genres_data = get_movie_path("genre/movie/list",my_params={
        "language" : "ko-kr",
    })
    genres = genres_data.get('genres')

    my_movie_list = []

    for movie in movies :
        movie_genre_id = movie.get('genre_ids')

        movie_genre_list = []
        for genre_id in movie_genre_id :
           #영화가 가지고 있는 각각의 id에 대해서
           # genre와 매칭

            for genre in genres :
                if genre['id'] == genre_id :
                    movie_genre_list.append(genre['name'])
        my_movie_list.append((movie.get('title'), movie_genre_list))

    pprint(my_movie_list)


except Exception as e:
    print(e)
```
