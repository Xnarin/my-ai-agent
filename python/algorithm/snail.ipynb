```python
# 범위를 벗어난 index -> 실행이 안되는게 맞는 코드입니다.

# 기초 정보
from pprint import pprint

n = 5
snail = [[0]*n for _ in range(n)]

i = 0
j = 0

# 방향성을 특정지어야 함.
#     우  하  좌  상
di = [0, 1, 0, -1]
dj = [1, 0, -1, 0]

# 우리의 진행 방향.
d = 0

# 1넣기 -> 숫자 바꾸고 이동하기
for num in range(1, n**2+1):
    snail[i][j] = num

    print()
    pprint(snail)

    # (0,n)에서 범위를 벗어남.
    # grid를 벗어난 경우 -> 방향을 바꿔줘야 함.
    # grid를 벗어난 경우 인덱스 관리가 어려워서
    # 내가 직접 이동하는게 아니라, (ni, nj)라는 이동의 후보군을 미리 보내놓습니다.

    ni = i + di[d]
    nj = j + dj[d]

    # ni, nj가 grid를 벗어나지 않아 -> 그냥 그대로 이동해
    # ni, nj가 grid를 벗어나 -> 방향 바꾸고 다시 이동해 -> 특별한 경우

    # if (ni, nj)가 grid를 벗어나 :
    # if ni < 0 or ni >= n or nj < 0 or nj >= n:

    # if not (ni, nj)가 grid 안에 있어
    if not ((0 <= ni < n) and (0 <= nj < n)):
        # 방향 이동
        d += 1
        # 새로 이동한 좌표
        print(d)
        ni = i + di[d]
        nj = j + dj[d]

    # 이동 확정
    i = ni
    j = nj
```

```python
# 방향을 나타내는 d를 변경

from pprint import pprint

n = 5
snail = [[0]*n for _ in range(n)]

i = 0
j = 0

# 방향성을 특정지어야 함.
#     우  하  좌  상
di = [0, 1, 0, -1]
dj = [1, 0, -1, 0]

# 우리의 진행 방향.
d = 0

# 1넣기 -> 숫자 바꾸고 이동하기
for num in range(1, n**2+1):
    snail[i][j] = num

    print()
    pprint(snail)


    ni = i + di[d]
    nj = j + dj[d]

    # 현재는 grid를 벗어났을 때에만 방향을 바꿈.
    # -> 이미 값이 있을때도 방향을 바꿈.
    # 즉, snail[ni][n]에 뭔가 있을 때, snail[ni][nj] != 0
    if not ((0 <= ni < n) and (0 <= nj < n)) or snail[ni][nj]:
        # d가 4가 되어버리면 index error 발생.
        # d가 4가 되면 0으로 바꿔줌.
        # d += 1
        # if d == 4:
        #     d = 0

        # 4로 나눈 나머지
        d = (d + 1) % 4

        ni = i + di[d]
        nj = j + dj[d]

    # 이동 
    i = ni
    j = nj
```
