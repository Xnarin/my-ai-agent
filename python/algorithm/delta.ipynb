2 와 근접해있는 1의 개수 구하기 -> 상하좌우 중 하나가 맞닿아 있을 때

```python
n, m = 5, 5

lst = [
    [1, 0, 1, 0, 2],
    [0, 0, 1, 2, 1],
    [0, 1, 0, 1, 2],
    [1, 1, 0, 2, 1],
    [2, 0, 1, 0, 2]
]

di = [0, 0, 1, -1]
dj = [1, -1, 0, 0]
count = 0

for i in range(n) :
    for j in range(m) :
        if(lst[i][j] == 1) :
            for d in range(4) :
                ni = i + di[d]
                nj = j + dj[d]

                if(0 <= ni <= n and 0 <= nj <= m) :
                    map = lst[ni][nj]

                    if(map == 2) : 
                        count += 1
                        break;

print(count)
```
