```python
from collections import deque

def solution(bridge_length, weight, truck_weights):
    answer = 0
    
    bridge = deque([0] * bridge_length)
    trucks = deque(truck_weights)
    
    current_weight = 0
    time = 0

    while trucks:
        time += 1

        # 다리에서 한 칸 이동
        current_weight -= bridge.popleft()

        # 다음 트럭이 올라갈 수 있으면 진입, 아니면 빈 공간(0) 진입
        if current_weight + trucks[0] <= weight:
            truck = trucks.popleft()
            bridge.append(truck)
            current_weight += truck
        else:
            bridge.append(0)

    # 마지막 트럭이 다리를 빠져나갈 시간
    return time + bridge_length
```

```python
from collections import deque 
def solution(s) :
    stack = deque()

    open_blanket = {"(", "{", "["}

    blanket_mapping = {
        "{" : "}",
        "[" : "]",
        "(" : ")"
    }

    for char in stack :
        if char in open_blanket :
            stack.append(char)

        elif char in blanket_mapping.keys() :
            if stack[-1] == blanket_mapping[char] :
                return False 
    return True
                
```
