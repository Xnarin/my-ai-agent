### for문 : 할당의 반복.  
같은 변수에 다른 값들을 반복적으로 할당합니다.

```python
names = ['jun', 'aiden', 'ken', 'alex']

for name in names:
    print(name)
print()

for name in ['jun', 'aiden', 'ken', 'alex']:
    # 0 : name = 'jun'
    # 1 : name = 'aiden'
    print(name)
```

```python
scores = [50, 60, 70, 100]

for score in scores:
    # score = 50
    print(score)
    # 만약 100점이면 축하해주자.
    if score == 100:
        print('축하해!')
```

# range
range() : 수열 만드는 함수스러운 것

```python
# range(4) -> 0, 1, 2, 3
# 0부터 시작해서 n개까지의 숫자를 만들어 냅니다.

for i in range(4):
    print(i)
```

```python
# 모두의 이름을 출력하고 싶다.
names = ['jun', 'aiden', 'ken', 'alex']
#          0        1      2       3

print(names[0])
print(names[1])
print(names[2])
print(names[3])
```

```python
names = ['jun', 'aiden', 'ken', 'alex', 'beemo']
#          0      1         2        3       4
for i in [0, 1, 2, 3]:
    print(names[i])

print()

for i in range(4):
    print(names[i])
```

```python
names = ['jun', 'aiden', 'ken', 'alex', 'beemo']

for i in range(len(names)):
    print(names[i])

# range(n) : 0부터 n개까지의 수열을 나타냄. 0, 1, 2, 3, ..., n-1
# len(list) : list의 길이를 나타냄
# range(len(list)) : 0부터 list의 길이까지의 수열을 나타냄. 따라서 index가 나옴.
```

# while


```python
# while 조건문 :
#     조건이 True일 때 실행.
#     반복되면서 조건문을 언젠가는 False로 바꿔야 함.


n = 0

while n < 10:
    print(n)
    n += 1
```

# break / continue


```python
# break
# 반복을 멈추고 싶을 때

# 짝수가 나오면 멈춰

numbers = [1, 3, 2, 23, 25, 4, 11, 26]

for number in numbers:

    print(number)
    if number % 2 == 0:
        break


    
```

```python

# n이 4보다 작으면 실행해.
n = 0

while n < 4:
    print(n)
    n += 1

print()
# 계속 실행 해. n이 4보다 커지면 멈춰.
n = 0

while True: # 잘못하면 무한루프가 됩니다.
    if n >= 4:
        break

    print(n)
    n += 1
```

```python
# continue
# 반복 중에 다음 반복으로 넘어가 싶을 때.

numbers = [1, 3, 2, 23, 25, 4, 11, 26]

# 출력해. 근데 짝수가 나오면 아무것도 하지 말고 넘어가.

for number in numbers:
    if number % 2 == 0:
        continue
    
    print(number)
```

```python
n = 0
while n < 20:
    n += 1
    if n % 3:
        continue
    print(n)
```
