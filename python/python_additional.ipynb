### `*args`, `**kwargs`

- `*args`: 여러 개의 위치 인자를 튜플로 받는다.
- `**kwargs`: 여러 개의 키워드 인자를 딕셔너리로 받는다.

함수를 정의할 때는 여러 인자를 하나로 모으고, 함수를 호출할 때는 `*`, `**`로 묶인 값을 펼친다.

```python
def show_info(*args, **kwargs):
    print(args)
    print(kwargs)

show_info(10, 20, name="철수", age=25)
```

### 패킹과 언패킹

패킹은 여러 값을 하나로 묶는 것이고, 언패킹은 묶인 값을 다시 펼치는 것이다. 함수 호출에서 `*`는 리스트나 튜플을, `**`는 딕셔너리를 펼친다.

```python
numbers = (10, 20, 30)  # 패킹
a, b, c = numbers       # 언패킹
print(a, b, c)

def add(x, y):
    return x + y

values = [10, 20]
print(add(*values))

def introduce(name, age):
    print(name, age)

user = {"name": "영희", "age": 20}
introduce(**user)
```

### 리스트의 얕은 복사와 깊은 복사

`b = a`는 복사가 아니라 두 변수가 같은 리스트를 가리키는 대입이다.

- 얕은 복사: 바깥 리스트만 새로 만들고 내부 리스트는 공유한다.
- 깊은 복사: 내부 리스트까지 모두 새로 만든다.

### `with` 문법

`with`를 사용하면 파일 사용이 끝난 뒤 자동으로 파일을 닫아 준다. 파일을 처리하다가 오류가 발생해도 안전하게 닫힌다.

```python
with open("example.txt", "w", encoding="utf-8") as file:
    file.write("안녕하세요")

with open("example.txt", "r", encoding="utf-8") as file:
    content = file.read()

print(content)
```

### `__str__`과 `__repr__`

- `__str__`: 사용자가 보기 편한 문자열이다. 객체를 `print()`하면 주로 이 메서드가 사용된다.
- `__repr__`: 개발자가 객체의 정보를 확인하기 위한 문자열이다.

```python
class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price

    def __str__(self):
        return f"{self.name} - {self.price}원"

    def __repr__(self):
        return f"Product({self.name!r}, {self.price})"

product = Product("키보드", 50000)
print(str(product))
print(repr(product))
```

### `is`와 `==`

- `==`: 두 객체의 값이 같은지 비교한다.
- `is`: 두 변수가 같은 객체를 가리키는지 비교한다.

숫자나 문자열 같은 값 비교에는 `is`를 사용하지 않고 `==`를 사용한다. `None` 비교에는 `is None`을 사용한다.

```python
a = [1, 2, 3]
b = [1, 2, 3]
c = a

print(a == b)  # True: 값이 같다.
print(a is b)  # False: 서로 다른 객체이다.
print(a is c)  # True: 같은 객체이다.

value = None
print(value is None)
```

```python
import copy

original = [[1, 2], [3, 4]]
shallow = original.copy()
deep = copy.deepcopy(original)

original[0].append(100)

print("원본:", original)
print("얕은 복사:", shallow)
print("깊은 복사:", deep)
```

### 제네릭

타입 힌트는 코드가 기대하는 자료형을 보여 주지만, 실행 중에 자료형을 강제로 검사하지는 않는다. 제네릭은 하나의 함수나 클래스를 여러 자료형에 사용할 수 있게 한다.

```python
from typing import TypeVar

T = TypeVar("T")

def get_first(items: list[T]) -> T:
    return items[0]

print(get_first([1, 2, 3]))
print(get_first(["가", "나", "다"]))
```

### 덕 타이핑과 `Protocol`

덕 타이핑은 클래스의 종류보다 필요한 기능이 있는지를 보는 방식이다. `Protocol`은 필요한 메서드를 타입 힌트로 나타낸다.

#### ABC 대신 덕 타이핑을 사용하는 이유

ABC(Abstract Base Class)는 공통 부모 클래스를 상속하게 하여 구현 규칙을 강제한다. 반면 덕 타이핑은 상속 관계가 없어도 필요한 메서드만 가지고 있으면 사용할 수 있다.

- 클래스 사이의 결합도가 낮아진다.
- 기존 클래스나 외부 라이브러리의 클래스를 수정하지 않고 사용할 수 있다.
- 서로 관련 없는 클래스도 같은 기능을 제공하면 동일하게 처리할 수 있다.
- `Protocol`을 함께 사용하면 덕 타이핑의 유연성을 유지하면서 정적 타입 검사도 받을 수 있다.

공통 구현을 재사용하거나 실행 중에 상속 규칙을 강제해야 한다면 ABC가 적합하고, 상속 관계보다 객체가 제공하는 기능이 중요하다면 덕 타이핑이 적합하다.

```python
from typing import Protocol

class Speaker(Protocol):
    def speak(self) -> str:
        pass

class Dog:
    def speak(self) -> str:
        return "멍멍"

class Cat:
    def speak(self) -> str:
        return "야옹"

def make_sound(animal: Speaker):
    print(animal.speak())

make_sound(Dog())
make_sound(Cat())
```

### 제너레이터

제너레이터는 모든 결과를 한꺼번에 메모리에 올리지 않고, **필요할 때마다 값을 하나씩 생성**하여 반환하는 도구이다.

- 일반 함수(`return`): 계산이 끝나면 결과를 반환하고 종료한다.
- 제너레이터(`yield`): 값을 하나씩 반환하고 실행을 일시 정지한다.

#### 리스트를 만들어 순회하는 대신 제너레이터를 사용하는 이유

일반적인 `for` 문은 리스트와 같은 반복 가능한 객체에서 값을 꺼내 사용한다. 제너레이터는 `for` 문을 대체하는 것이 아니라, `for` 문에 값을 하나씩 제공하는 반복 가능한 객체를 만드는 방식이다.

- 모든 값을 미리 저장하지 않아 메모리를 적게 사용한다.
- 첫 번째 값을 만든 즉시 처리를 시작할 수 있어 대기 시간이 줄어든다.
- 파일 읽기, 스트리밍 데이터, 무한 수열처럼 전체 크기를 알 수 없는 데이터도 처리할 수 있다.
- 필요한 시점에만 값을 계산하므로 사용하지 않는 값에 대한 연산을 줄일 수 있다.

데이터가 작고 여러 번 반복해서 사용해야 한다면 리스트가 편리하다. 데이터가 크거나 한 번만 순서대로 처리한다면 제너레이터가 적합하다.

```python
def simple_generator():
    print("첫 번째 값을 생성합니다.")
    yield "A"

    print("두 번째 값을 생성합니다.")
    yield "B"

    print("세 번째 값을 생성합니다.")
    yield "C"

gen = simple_generator()

# next()를 호출할 때마다 다음 yield까지 실행된다.
print(f"받은 값: {next(gen)}")
print(f"받은 값: {next(gen)}")
print(f"받은 값: {next(gen)}")
```

```python
gen = simple_generator()

# for 문은 내부적으로 next()를 호출한다.
for text in gen:
    print(text)
```
