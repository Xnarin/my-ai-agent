```python
from abc import ABC, abstractmethod
```

```python
class Character(ABC) : 
    def __init__(self, name, role) :
        self.name = name
        self.role = role
        self.is_handed = False
        print(f"[{self.role}] {self.name} 탄생합니다.")
        
    @abstractmethod
    def attack(self) :
        pass

    @abstractmethod
    def hand_weapon(self, weapon) :
        pass
    
    @abstractmethod
    def finish_move(self):
        pass
```

```python
class Warrior(Character) : 

    def attack(self) :
        print(f"{self.name}, 전사답게 검을 휘두르며 달려나갑니다.")

    def hand_weapon(self, weapon) :
        if(weapon == "검") :
            self.weapon = weapon
            self.is_handed = True
            print(f"{self.name}, 이건 제가 잘 다루는 거죠라며 {weapon}을 장착합니다")
        else : print(f"{self.name}, 이건 내 역할에 맞지 않아요라며 {weapon}를 거절합니다.") 

    def finish_move(self):
        if(self.is_handed) :
            print(f"{self.name}, 필살기 연속 횡베기를 사용합니다.")

        else : print(f"{self.name}, 무기가 없어 필살기를 사용할 수 없습니다.")
```

```python
class Wizard(Character) : 

    def attack(self) :
        print(f"{self.name}, 마법사답게 파이어볼을 씁니다.")

    def hand_weapon(self, weapon) :
        if(weapon == "지팡이") :
            self.weapon = weapon
            self.is_handed = True
            print(f"{self.name}, 이건 제가 잘 다루는 거죠라며 {weapon}을 장착합니다")
        else : print(f"{self.name}, 이건 내 역할에 맞지 않아요라며 {weapon}를 거절합니다.") 

    def finish_move(self):
        if(self.is_handed) :
            print(f"{self.name}, 필살기 메테오를 사용합니다.")

        else : print(f"{self.name}, 무기가 없어 필살기를 사용할 수 없습니다.")
```

```python
c1 = Warrior("아더", "전사")
```

```python
c1.attack()
c1.hand_weapon("검")
c1.finish_move()
```

```python
c2 = Wizard("사샤", "마법사")
c2.attack()
c2.finish_move()
c2.hand_weapon("검")
c2.hand_weapon("지팡이")
c2.finish_move()
```
