### 1. 요구 사항
#### [스마트 허브(SmartHub) 설계]
- 허브는 이름과 지원하는 연결 방식(Protocol)을 속성으로 가집니다.
- 허브는 여러 개의 기기(Device)를 저장할 수 있는 리스트를 가집니다.
- 기능: 기기_등록() 전체가동(), 전체_중지()

```python
from abc import ABC, abstractmethod
```

```python
class Protocol(ABC):
    @abstractmethod
    def protocol_on(self):
        pass
class WiFiProtocol(Protocol) :
    def protocol_on(self):
        print("WiFi 망을 구성합니다.")
class ZigbeeProtocol(Protocol) :
    def protocol_on(self):
        print("Zigbee 망을 구성합니다.")
class BluetoothProtocol(Protocol) :
    def protocol_on(self):
        print("Bluetooth 망을 구성합니다.")
```

```python
class Device(ABC):
    
    def __init__(self, name, brand, protocol):
        self.name = name
        self.protocol = protocol
        self.brand = brand
   
    @abstractmethod
    def work_on(self):
        pass

    @abstractmethod
    def status(self) :
        pass 
    
    @abstractmethod
    def work_off(self):
        pass
```

```python
class SmartLight(Device):
    def work_on(self):
        print(f"{self.name} 전등을 켭니다.")

    def status(self):
        print(f"{self.name} 전등 상태를 확인합니다.")

    def work_off(self):
        print(f"{self.name} 전등을 끕니다.")


class AirConditioner(Device):
    def work_on(self):
        print(f"{self.name} 에어컨을 켭니다.")

    def status(self):
        print(f"{self.name} 에어컨 상태를 확인합니다.")

    def work_off(self):
        print(f"{self.name} 에어컨을 끕니다.")
```

```python
class SmartHub :
    def __init__(self, name, protocol) :
        self.name = name 
        self.protocol = protocol
        self.device_lst = []

    def register(self, device : Device) :
        if(type(device.protocol) != type(self.protocol)):
            print("호환되지 않는 연결 방식입니다")
        else : self.device_lst.append(device)
    def all_on(self) : 
        for device in self.device_lst :
            device.protocol.protocol_on()
            device.work_on()
    def all_off(self) :
        print(f"{self.name} 전체 중지")
        for device in self.device_lst :
            device.work_off()
```

```python
xiaomi_hub = SmartHub("Xiaomi Hub", ZigbeeProtocol())

philips_light = SmartLight("Philips Light", "Philips", WiFiProtocol())
ikea_light = SmartLight("IKEA Light", "IKEA", ZigbeeProtocol())

xiaomi_hub.register(philips_light)  # Fails
xiaomi_hub.register(ikea_light)     # Succeeds

xiaomi_hub.all_on()
xiaomi_hub.all_off()
```
