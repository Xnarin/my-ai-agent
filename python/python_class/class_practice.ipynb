```python
class Person:

    # 클래스 변수
    # 클래스의 모든 인스턴스들이 공유하는 값.
    population = 0

    def __init__(self, name, age):
        self.name = name
        self.age = age

        Person.population += 1

    def introduce(self):
        print(f'나의 이름은 {self.name}, 나이는 {self.age}')

    def birthday(self):
        self.age += 1

# p1은 Person 클래스의 인스턴스
# p1은 객체.
p1 = Person('김철수', 25)

print(p1.name)
p1.introduce()
print('---------------')
print(p1.age)
p1.birthday()
print(p1.age)
p1.age += 1
print(p1.age)

print('---------------')

print(Person.population)

p2 = Person('영희', 23)
print('영희 탄생')

print(Person.population)
print(p1.population)
print(p2.population)

print('---------------')
p1.population = 100 # p1의 인스턴스 변수에 populaion을 추가한 것.
print(p1.population)
print(p2.population)
```

```python
class Person:
    def __init__(self, name, hobbies):
        self.name = name
        self.hobbies = hobbies

    def __str__(self):
        # 객체를 출력하거나 문자열로 변환할 때 반환할 값 정의
        return f"이름: {self.name}"

    def __len__(self):
        # len 함수 호출 시 반환할 정수 값 정의
        # return len(self.hobbies)

        return len(self.name)

p = Person("세종대왕", ["독서", "음악", "연구"])

print(p.name)
print(p.hobbies)
# # __str__ 호출 결과 출력
print(p)
# # __len__ 호출 결과 출력
print(len(p))

```
