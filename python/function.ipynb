```python
# 길이를 재는 기능

words = ['a', 'b', 'c', 'd']

count = 0
for _ in words:
    count += 1

# 길이를 재는 함수

# input : 길이를 잴 수 있는 무언가 -> list.
# 기능 : 길이를 잰다. 즉 원소의 개수를 센다.
# output : 길이.

def my_len(lst: list) -> int:

    count = 0
    for _ in lst:
        count += 1    

    return count

count = my_len(words)

print(count)
```

```python
orders1 = [
    {
        "order_id": "ORD-001",
        "items": [
            {"name": "사과", "status": "in_stock"},
            {"name": "배", "status": "in_stock"}
        ]
    },
    {
        "order_id": "ORD-002",
        "items": [
            {"name": "포도", "status": "out_of_stock"},
            {"name": "딸기", "status": "in_stock"}
        ]
    },
    {
        "order_id": "ORD-003",
        "items": [
            {"name": "수박", "status": "out_of_stock"}
        ]
    },
    {
        "order_id": "ORD-004",
        "items": [
            {"name": "참외", "status": "out_of_stock"},
            {"name": "멜론", "status": "out_of_stock"}
        ]
    },
    {
        "order_id": "ORD-005",
        "items": [
            {"name": "망고", "status": "out_of_stock"},
            {"name": "토마토", "status": "in_stock"},
            {"name": "호두", "status": "out_of_stock"}
        ]
    },
]

orders2 = []

orders3 = []


def func(orders):
    out_of_stock_items = []
    out_of_stock_orders = []
    for order in orders:

        order_id = order['order_id']
        items = order['items']

        has_out_of_stock = False 
        # 상태를 items를 돌면서 그때그때 바꾸는게 아니라, 하나의 order에 대해서 변수로 활용합니다.
        for item in items:
            name = item['name']
            status = item['status']
            if status == 'out_of_stock':
                has_out_of_stock = True
                out_of_stock_items.append(name)

        if has_out_of_stock:
            out_of_stock_orders.append(order_id)

    return out_of_stock_items, out_of_stock_orders



func(orders1)



func(orders2)



func(orders3)



```
