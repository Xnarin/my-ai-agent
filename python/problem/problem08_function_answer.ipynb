2. 점수가 60점 미만이면 "불합격"을 반환합니다.

```python
# 
score = -1
if score < 0 or score > 100:
    print('잘못된 점수')
elif score >= 60:
    print('합격')
elif score <= 60:
    print('불합격')

# input : score. 정수
# 기능 : 위에서 만듦.
# output : 위에서는 print하였지만, 함수에서는 return.
```

```python
def get_grade_result(score):
    if score < 0 or score > 100:
        return '잘못된 점수'
    elif score >= 60:
        return '합격'
    elif score <= 60:
        return '불합격'
    
```

```python
# return은 함수를 끝낸다.
# 따라서 elif 에서 el 을 삭제해도 같은 기능을 하게 되요.
# 반드시 함수의 if에서 return을 만나면 elif를 하면 안되냐? 아닙니다. 가독성이 중요.
# 아래는 그냥 그렇다는걸 알려주기 위한 예시.

def get_grade_result(score):
    if score < 0 or score > 100:
        return '잘못된 점수'
    
    if score >= 60:
        return '합격'
    elif score <= 60:
        return '불합격'
    
```

```python

# 테스트
print(get_grade_result(85))  # 결과: "합격"
print(get_grade_result(40))  # 결과: "불합격"
print(get_grade_result(120)) # 결과: "잘못된 점수"
```

2. 숫자가 홀수이면 "홀수"를 반환합니다.

```python

# 삼항연산자
# a if 조건 else b
# 조건이 True면 a, False면 b 
def check_even_odd(n):
    return '홀수' if n % 2 else '짝수'
```

### 문제 3
두 개의 숫자와 연산자(+, -)를 입력받아 계산 결과를 반환하는 간단한 계산기 함수 `simple_calc`를 작성하세요.

**조건:**
1. `op`가 "+"이면 두 수의 합을 반환합니다.
2. `op`가 "-"이면 두 수의 차를 반환합니다.
3. 그 외의 연산자가 들어오면 "지원하지 않는 연산"이라는 메시지를 반환하세요.

```python
num1 = 10
num2 = 20
op = '+'

def simple_calc(num1, num2, op):
    if op == '+':
        return num1 + num2
    
    if op == '-':
        return num1 - num2
    
    return '지원하지 않는 연산'
```

```python

# 테스트
print(simple_calc(10, 5, "+")) # 결과: 15
print(simple_calc(10, 5, "-")) # 결과: 5
print(simple_calc(10, 5, "*")) # 결과: "지원하지 않는 연산"
```

```python
name_list = ["김철수", "이영희", "박씨", "최도날드"]

# input : 이름 리스트
# 기능  이름이 3자 이상인 사람만 고르자.
# output : 새로운 리스트를 반환해라.

long_name_list = []
for name in name_list:
    if len(name) >= 3:
        long_name_list.append(name)
print(long_name_list)

```

```python
def filter_long_names(name_list):
    long_name_list = []
    for name in name_list:
        if len(name) >= 3:
            long_name_list.append(name)
    return long_name_list
```

```python
def my_len(iter):
    count = 0
    for _ in iter:
        count += 1
    return count

def filter_long_names(name_list):
    long_name_list = []
    for name in name_list:
        if my_len(name) >= 3:
            long_name_list.append(name)
    return long_name_list
```

```python

# 테스트
name_list = ["김철수", "이영희", "박씨", "최도날드"]
print(filter_long_names(name_list)) # 결과: ["김철수", "이영희", "최도날드"]
# print([name for name in name_list if len(name) >= 3])
```

### 문제 7
서버에서 받아온 다양한 형태의 데이터가 섞인 리스트에서 정수 ID만 추출하여 정형화하는 함수 `normalize_user_data`를 작성하세요.

**조건:**
1. 숫자형 문자열("105"), 숫자(101), 리스트 안의 숫자([102]) 등을 모두 정수로 변환하세요.
2. 중복된 ID는 제거하고, 결과는 오름차순으로 정렬된 리스트여야 합니다.

```python
dirty_data = ["105", 101, [102, 103], "101", 104, [105]]

# dirty_data = ["105", 101, "101", 104, ]

def normalize_user_data(dirty_data):
    clean_data = []

    for data in dirty_data:
        # 1. 숫자인 경우 -> 그냥 저장한다.
        # 2. 숫자형 문자열 -> int로 변환 후 저장한다.

        # 3. 리스트 안의 숫자 -> 반복을 돌면서 저장한다.
        if type(data) == list:
            for datum in data:
                clean_data.append(datum)
        else:
            clean_data.append(int(data))

    clean_data = set(clean_data) # 중복제거
    clean_data = sorted(clean_data)
    return clean_data

        

```

```python
dirty_data = ["105", 101, [102, 103], "101", 104, [105]]

# dirty_data = ["105", 101, "101", 104, ]

def normalize_user_data(dirty_data):

    clean_data = set() # 문제 조건에서 중복 제거가 있기 때문.

    for data in dirty_data:
        # 3. [123, 456] -> 숫자의 리스트.
        if type(data) == list: # isinstance
            for num in data:
                clean_data.add(num)

        # 1. "123"
        # 2. 123
        else:
            clean_data.add(int(data))

    clean_data = sorted(clean_data)
    return clean_data

```

```python


dirty_data = ["105", 101, [102, 103], "101", 104, [105]]
print(normalize_user_data(dirty_data)) # 결과: [101, 102, 103, 104, 105]
```

### 문제 8
특정 조건을 만족하는 유저들을 추출하여 알림 대상자 리스트를 반환하는 함수 `get_notification_targets`를 작성하세요.

**조건:**
1. `is_active`가 `True`인 유저 중에서 선별합니다.
2. `last_login`이 7일 이상이거나, `point`가 10,000점 이상인 유저가 대상입니다.
3. 대상자의 'name'만 리스트에 담아 반환하세요.

```python
users_data = {
    "u1": {"name": "Alice", "is_active": True, "last_login": 10, "point": 5000},
    "u2": {"name": "Bob", "is_active": False, "last_login": 1, "point": 15000},
    "u3": {"name": "Charlie", "is_active": True, "last_login": 2, "point": 12000}
}

def get_notification_targets(users_data):
    user_list = []

    for user_data in users_data.values():
        is_active = user_data['is_active']
        name = user_data['name']
        last_login = user_data['last_login']
        point = user_data['point']


        # 1. `is_active`가 `True`인 유저 중에서 선별합니다.
        if not is_active:
            continue

        # 2. `last_login`이 7일 이상이거나, `point`가 10,000점 이상인 유저가 대상입니다.
        if not (last_login >= 7 or point >= 10000):
            continue

        # 3. 대상자의 'name'만 리스트에 담아 반환하세요.
        user_list.append(name)
    return user_list
        
        


```

```python


users_data = {
    "u1": {"name": "Alice", "is_active": True, "last_login": 10, "point": 5000},
    "u2": {"name": "Bob", "is_active": False, "last_login": 1, "point": 1500},
    "u3": {"name": "Charlie", "is_active": True, "last_login": 2, "point": 12000}
}
print(get_notification_targets(users_data)) # 결과: ['Alice', 'Charlie']
```
