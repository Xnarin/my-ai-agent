### 문제 1
다음 리스트를 만들어보세요 (여러가지 방법으로 만들어보세요)
- a. [0, 1, 2, 3, 4, 5]
- b. [4, 5, 6, 7]
- c. [3, 6, 9]
- d. [1, 2, 1, 2, 1, 2, 1, 2]

```python
a = [0, 1, 2, 3, 4, 5]

#######

a = list(range(6))
print(a)

####### 
# 빈 공간 ([])이 존재하고, 거기에 0을 넣기 -> 1을 넣기 -> 2를 넣기/

a = []
for num in range(6):
    a.append(num)
print(a)

#######
a = [num for num in range(6)]
print(a)

####### 
a = []
num = 0

while num <= 5:
    a.append(num)
    num += 1
print(a)

####### 
a = []
num = -1

while num < 5:
    num += 1
    a.append(num)
print(a)

####### 
a = []
num = 0

while True:
    a.append(num)

    if num == 5:
        break
    
    num += 1
print(a)
```

```python
# - b. [4, 5, 6, 7]

##########
b = []
# 4, 5, 6, 7 -> 0+4, 1+4, 2+4, 3+4
for num in range(4):
    b.append(num + 4)
print(b)

##########
b = [num + 4 for num in range(4)]
print(b)

##########
b = []
num = 4
while num <= 7:
    b.append(num)
    num += 1
print(b)

##########
b = list(range(8))
print(b)
b = b[4:]
print(b)

##########
b = list(range(4, 8))
print(b)
```

```python
# - c. [3, 6, 9]
##########
# [0, 1, 2] -> [3, 6, 9] 
c = []
for num in range(3):
    c.append(3 * (num+1))
print(c)

##########
# 0+3 -> 3+3 -> 3+3+3
num = 0
c = []
while num < 9:
    num += 3
    c.append(num)
print(c)
```

```python
# - d. [1, 2, 1, 2, 1, 2, 1, 2]
##########
d = [1, 2] * 4

##########
d = []
for _ in range(4):
    d.append(1)
    d.append(2)
print(d)

##########
d = []
for _ in range(4):
    # d.append(1)
    # d.append(2)
    # d.append(3)
    # d.append(4)
    for num in range(4):
        d.append(num + 1)
print(d)
```

### 문제 2
구구단 3단을 출력하시오. 단, for, while을 활용하여 다양하게 만들어보시오.

```python
# 3x1, 3x2... 3x9
n = 3 # n단
m = 9 # 몇개까지 할껀지. 구구단 / 1919단

for i in range(m):
    print(n * (i+1), end=" ")

####################################
print()
num = 0
count = 0
while count < m:
    num += n
    count += 1
    print(num, end=" ")
```

### 문제 3
40 이하의 3의 배수를 출력하시오

```python
n = 40
for num in range(1, n + 1):
    if num % 3 == 0:
        print(num, end=" ")

####################################
print()
num = 3
while num < 40:
    print(num, end=" ")
    num += 3

####################################
print()
for i in range(n//3): # 구구단 코드 재활용 -> 40 이하의 3의 배수 개수는 3으로 나눈 몫
    print(3 * (i+1), end=" ")
```

### 문제 6
1 ~ 9 까지의 자연수 중 제곱한 수가 10 이상 50 이하인 자연수의 리스트를 출력해보세요.

### 문제 7
숫자 두개를 각각 입력받아 두 수 사이에 속한 모든 정수의 합을 구하시오. (ex. 2와 5를 입력받는다면 2 + 3 + 4 + 5 = 14)

```python
a = 2
b = 5
# 2 + 3 + 4 + 5 = 14

num = a
total = 0
while num <= b:
    total += num
    num += 1

print(total)
```

```python
# 만약 a가 더 크다면? -> a를 더 작게 바꿔주면 된다.
a = 5
b = 2
# a = 5 -> 2
# b = 2 -> 5

# a, b = b, a # pythonic한 방법
if a > b:
    tmp = a
    a = b
    b = tmp

print('a : ', a,'b : ', b)

num = a
total = 0
while num <= b:
    total += num
    num += 1

print(total)
```

### 문제 9
1부터 50까지의 숫자 중에서 5의 배수만 골라 `my_list`에 저장하고, 그 리스트의 모든 요소의 합을 `for`문을 사용하여 구하세요.

```python
my_list = []
for num in range(1, 51):
    if num % 5 == 0:
        my_list.append(num)

print(my_list)

total = 0
for num in my_list:
    total += num
print(total)
```

### 문제 11
사용자로부터 숫자를 계속 입력받다가 `-1`을 입력하면 입력을 멈추고, 그동안 입력받은 숫자들 중에서 **가장 큰 값**을 출력하는 프로그램을 `while`문을 사용하여 작성하세요. (단, `max()` 함수를 사용하지 마세요.)

```python
# 입력 받기
while True:
    input_num = int(input())
    print(input_num)

    if input_num == -1:
        break
```

```python

# 최대값 찾기
lst = [1, 3, 2, 5, 4]
max_value = lst[0]

for num in lst:
    if num > max_value:
        max_value = num
print(max_value)
```

```python
# 입력 받기
lst = []
while True:
    input_num = int(input())
    # print(input_num)
    lst.append(input_num)

    if input_num == -1:
        break
print(lst)

# 최대값 찾기
# lst = [1, 3, 2, 5, 4]
max_value = lst[0]

for num in lst:
    if num > max_value:
        max_value = num
print(max_value)
```

```python
# lst를 만들면서 진행하는 것이 아니라
# 숫자를 받으면서 하나하나 그때그때 계산하면 더 간단함.

# 단, 입력하는 숫자는 양의 정수.
# -1로 멈추기 때문에 -2 등이 들어가면 깔끔하지 못함.

max_value = 0
while True:
    input_num = int(input())
    print(input_num)

    if input_num == -1:
        break

    if input_num > max_value:
        max_value = input_num
print(max_value)
```

### 문제 12
`while`문을 사용하여 1부터 100까지의 정수 중 홀수의 합과 짝수의 합을 각각 따로 구하여 출력하세요.

```python
num = 1
odd_total = 0
even_total = 0

while num <= 100:

    print(num)
    if num % 2:
        odd_total += num
    else:
        even_total += num

    num += 1
    
print(odd_total, even_total)
```

### 문제 13
n = 17일 때, n이 소수인지 판단하시오. 

(추가로, n을 입력받아 같은 과정을 반복하시오.)

```python
# 짝수 -> 
# 2, 4, 6, 8, 10 -> 
# 2의 배수 -> 
# 2로 나눈 나머지가 0인 녀석들이네 -> 
# num % 2 == 0

# 소수
# 2, 3, 5, 7, 11, 13, 17, 19
# 약수가 없음 / 자기 자신과 1로만 나누어 지는 것.

# 자기 자신 or 1로 나누었을 때에만! 나머지가 0인 것. -> 소수이다.

# 자기 자신 or 1을 제외한 것들로 나누었을 때 나머지가 존재하는 것 -> 소수가 아니다.

# 소수를 구하기 위해서는 "모든 값"들에 대해서 나머지가 존재하는지를 확인해야 함.
# 소수가 아님을 구하기 위해서는 "어떤 값- 1개면 충분."들에 대해서 나머지가 존재하면 됩니다.


# 너는 소수가 맞아. 라고 가정.
# 나머지가 0인 녀석이 발견된다? -> 소수가 아니야! 로 변경할껍니다.

n = 17
is_prime = True
for num in range(2, n):
    if n % num == 0:
        # 약수가 존재하는거니까
        is_prime = False
        break
print(is_prime)
```
