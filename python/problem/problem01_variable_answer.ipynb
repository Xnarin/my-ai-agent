### 문제 5
물건의 가격이 price이고, 낸 돈이 money일 때, 거스름돈에서 500원 동전의 개수와 100원 동전의 개수를 구하는 프로그램을 작성하세요. 
(단, 가격은 항상 100원 단위이며 money > price라고 가정합니다.)

예: price = 3200, money = 5000 -> 500원: 3개, 100원: 3개

````python
price = 3100
money = 5000

# 거스름돈에 대한 변수 만들기.
change = money - price

count_500 = change // 500
remainder = change % 500


count_100 = remainder // 100

print(count_500)
print(remainder)
print(count_100)
````

````python
count_500, remainder = divmod(change, 500)
print(count_500, remainder)
````
