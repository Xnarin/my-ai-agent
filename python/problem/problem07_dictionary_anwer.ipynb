### 문제 1
API로 전달받은 게시글 데이터입니다. `status`가 'published'인 게시글들만 필터링하여, 해당 게시글들의 `view_count` 총합을 구하세요.

```python
posts = [
    {'title': '첫 번째 글', 'status': 'published', 'view_count': 120},
    {'title': '두 번째 글', 'status': 'draft', 'view_count': 0},
    {'title': '세 번째 글', 'status': 'published', 'view_count': 350},
    {'title': '네 번째 글', 'status': 'deleted', 'view_count': 10}
]

total_view_count = 0

for post in posts:
    # print(post)
    status = post['status']
    # print(status)

    if status == 'published':
        # print(status)
        view_count = post['view_count']

        total_view_count += view_count

print(total_view_count)
```

### 문제 2
유저 권한 시스템 로직입니다. 아래 규칙에 따라 접근 가능 여부(`True`/`False`)를 판단하여 결과를 적절한 형태로 만드세요.
1. `role`이 'admin'이면 점수와 상관없이 `True`
2. `role`이 'user'이고 `score`가 80점 이상이면 `True`
3. 그 외(guest 등)는 모두 `False`

```python
users = [
    {'name': 'Kim', 'info': {'role': 'admin', 'score': 50}},
    {'name': 'Lee', 'info': {'role': 'user', 'score': 85}},
    {'name': 'Park', 'info': {'role': 'user', 'score': 70}},
    {'name': 'Choi', 'info': {'role': 'guest', 'score': 100}}
]
users = [
    {
        'name': 'Kim', 
        'info': {
                    'role': 'admin', 
                    'score': 50
                }
    },
    {'name': 'Lee', 'info': {'role': 'user', 'score': 85}},
    {'name': 'Park', 'info': {'role': 'user', 'score': 70}},
    {'name': 'Choi', 'info': {'role': 'guest', 'score': 100}}
]

# 조건에 따라 name을 access_results에 넣고 싶다.
access_results = []

for user in users:
    # print(user)
    status = False # 기본값

    name = user['name']
    # print(name)

    info = user['info']
    # print(info)

    role = info['role']
    # role = user['info']['role']
    score = info['score']

    # 1. `role`이 'admin'이면 점수와 상관없이 `True`
    if role == 'admin':
        status = True


    # 2. `role`이 'user'이고 `score`가 80점 이상이면 `True`

    elif role == 'user' and score >= 80:
        status = True

    # 3. 그 외(guest 등)는 모두 `False`
    # 기본값이 False니까 냅둡니다.
    
    if status:
        access_results.append(name)


# 코드를 작성하세요

print(f"접근 권한 결과: {access_results}")
```

```python
users = [
    {'name': 'Kim', 'info': {'role': 'admin', 'score': 50}},
    {'name': 'Lee', 'info': {'role': 'user', 'score': 85}},
    {'name': 'Park', 'info': {'role': 'user', 'score': 70}},
    {'name': 'Choi', 'info': {'role': 'guest', 'score': 100}}
]
users = [
    {
        'name': 'Kim', 
        'info': {
                    'role': 'admin', 
                    'score': 50
                }
    },
    {'name': 'Lee', 'info': {'role': 'user', 'score': 85}},
    {'name': 'Park', 'info': {'role': 'user', 'score': 70}},
    {'name': 'Choi', 'info': {'role': 'guest', 'score': 100}}
]

# 조건에 따라 name을 access_results에 넣고 싶다.
access_results = []

for user in users:
    status = False # 기본값

    name = user['name']
    info = user['info']

    role = info['role']
    score = info['score']

    # 질문 : 두 if문을 합치면 안되나요? -> 됩니다. 단.
    # 개인적으로는 두개의 다른 조건이기 때문에 굳이 하나의 if문에서 다루기 보다는
    # 따로 분리해서 보는게 더 나아 보입니다.

    # if role == 'admin' or (role == 'user' and score >= 80): # 아래보다는 이게 낫구요.
    if role == 'admin' or role == 'user' and score >= 80:
        status = True


    if status:
        access_results.append(name)


# 코드를 작성하세요

print(f"접근 권한 결과: {access_results}")
```

### 문제 3
쇼핑몰의 카테고리별 상품 데이터입니다. 전체 상품 중 가격(`price`)이 50,000원 이상인 상품의 이름(`name`)만 모아서 하나의 리스트로 만드세요.

```python
store_data = {
    "electronics": [
        {"name": "키보드", "price": 45000},
        {"name": "모니터", "price": 200000}
    ],
    "fashion": [
        {"name": "티셔츠", "price": 20000},
        {"name": "청바지", "price": 60000},
        {"name": "재킷", "price": 120000}
    ],
    "furniture": [
        {"name": "의자", "price": 200000},
        {"name": "책상", "price": 600000},
        {"name": "책장", "price": 1200000}
    ]
}

expensive_product = []
for products in store_data.values():
    # products라는 상품의 리스트가 존재한다.
    
    for product in products:
        # print(product)
        name = product['name']
        price = product['price']
        print(name, price)

        if price >= 50000:
            expensive_product.append(name)
print(expensive_product)

```

```python
# 전체 상품 중 이 아니라 카테고리 별 비싼 상품을 모으세요!

store_data = {
    "electronics": [
        {"name": "키보드", "price": 45000},
        {"name": "모니터", "price": 200000}
    ],
    "fashion": [
        {"name": "티셔츠", "price": 20000},
        {"name": "청바지", "price": 60000},
        {"name": "재킷", "price": 120000}
    ],
    "furniture": [
        {"name": "의자", "price": 200000},
        {"name": "책상", "price": 600000},
        {"name": "책장", "price": 1200000}
    ]
}

expensive_product_by_category = {}

for category, products in store_data.items():
    # print(category, products)

    expensive_product = []

    for product in products:
        name = product['name']
        price = product['price']
        if price >= 50000:
            expensive_product.append(name)

    # print(category, expensive_product)
    expensive_product_by_category[category] = expensive_product

print(expensive_product_by_category)

```

### 문제 4
주문 내역 데이터입니다. 각 주문 별로 'items' 리스트가 들어있습니다. 전체 주문을 통틀어 '품절(out_of_stock)' 상태가 포함된 주문의 `order_id` 리스트를 구하세요. 또한 품절 상태인 items의 name에 대한 list를 구하세요.

```python
orders = [
    {
        "order_id": "ORD-001",
        "items": [
            {"name": "사과", "status": "in_stock"},
            {"name": "배", "status": "in_stock"}
        ]
    },
    {
        "order_id": "ORD-002",
        "items": [
            {"name": "포도", "status": "out_of_stock"},
            {"name": "딸기", "status": "in_stock"}
        ]
    },
    {
        "order_id": "ORD-003",
        "items": [
            {"name": "수박", "status": "out_of_stock"}
        ]
    },
    {
        "order_id": "ORD-004",
        "items": [
            {"name": "참외", "status": "out_of_stock"},
            {"name": "멜론", "status": "out_of_stock"}
        ]
    },
    {
        "order_id": "ORD-005",
        "items": [
            {"name": "망고", "status": "out_of_stock"},
            {"name": "토마토", "status": "in_stock"},
            {"name": "호두", "status": "out_of_stock"}
        ]
    },
]




```

```python

# 전체 주문을 통틀어 '품절(out_of_stock)' 상태가 포함된 주문의 `order_id` 리스트를 구하세요. 

out_of_stock_orders = []

for order in orders:
    # print(order)

    order_id = order['order_id']
    items = order['items']
    # print(items)

    for item in items:
        # print(item)

        status = item['status']
        if status == 'out_of_stock':
            # print('out_of_stock')
            out_of_stock_orders.append(order_id)
            break # 하나의 주문에서 여러개가 out_of_stock일 때 여러번 들어가는 것을 방지하는 목적.


print(f"품절 포함 주문 ID: {out_of_stock_orders}")
```

```python
# 또한 품절 상태인 items의 name에 대한 list를 구하세요.

out_of_stock_items = []

for order in orders:
    items = order['items']

    for item in items:
        
        name = item['name']
        status = item['status']
        # print(name, status)

        if status == 'out_of_stock':
            out_of_stock_items.append(name)
print(out_of_stock_items)
```

```python

# 합친버젼

out_of_stock_orders = []
out_of_stock_items = []

for order in orders:

    order_id = order['order_id']
    items = order['items']

    has_out_of_stock = False 
    # 상태를 items를 돌면서 그때그때 바꾸는게 아니라, 하나의 order에 대해서 변수로 활용합니다.

    for item in items:
        name = item['name']
        status = item['status']
        if status == 'out_of_stock':
            has_out_of_stock = True
            out_of_stock_items.append(name)

    if has_out_of_stock:
        out_of_stock_orders.append(order_id)



print(f"품절 포함 주문 ID: {out_of_stock_orders}")
print(f"품절 아이템: {out_of_stock_items}")
```

```python

# 합친버젼 - 조건에 따라 이미 있으면 안넣기.

out_of_stock_orders = []
out_of_stock_items = []

for order in orders:

    order_id = order['order_id']
    items = order['items']

    for item in items:
        name = item['name']
        status = item['status']
        if status == 'out_of_stock':
            out_of_stock_items.append(name)

            # order_id는 out_of_stock_orders에 없을때만 들어갈 수 있다.
            if order_id not in out_of_stock_orders:
                out_of_stock_orders.append(order_id)




print(f"품절 포함 주문 ID: {out_of_stock_orders}")
print(f"품절 아이템: {out_of_stock_items}")
```

```python

# 합친버젼 - 여러번 넣어도 되는 set 활용하기.

out_of_stock_orders = set()
out_of_stock_items = []

for order in orders:

    order_id = order['order_id']
    items = order['items']

    for item in items:
        name = item['name']
        status = item['status']
        if status == 'out_of_stock':
            out_of_stock_items.append(name)
            out_of_stock_orders.add(order_id)




print(f"품절 포함 주문 ID: {out_of_stock_orders}")
print(f"품절 아이템: {out_of_stock_items}")

print(list(out_of_stock_orders))
```
