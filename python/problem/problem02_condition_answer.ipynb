```python
year = 2000

if year % 4 == 0:
    print('윤년')
    if year % 100 == 0:
        print('평년')
        if year % 400 == 0:
            print('윤년')
else:
    print('평년')
```

### 문제 8
영화관 입장료는 기본 12,000원입니다. 다음 조건에 따라 최종 요금을 계산하여 출력하세요.
- 나이가 13세 미만이면 50% 할인
- 나이가 65세 이상이면 70% 할인
- '쿠폰'이 있는 경우 추가로 2,000원 할인

최종적으로 "당신의 입장료는 OO원입니다."라고 출력하는 코드를 작성하세요.

```python
age = 15
has_coupon = True

enter_fee = 12000

discount_rate_children = 0.5
discount_rate_elder = 0.7

discount_amount_coupon = 2000

if age < 13:
    enter_fee = enter_fee * (1 - discount_rate_children)
elif age >= 65:
    enter_fee = enter_fee * (1 - discount_rate_elder)

if has_coupon:
    enter_fee = enter_fee - discount_amount_coupon

print(enter_fee)
```
