
### 문제 1
사용자로부터 정수 `n`을 입력받아, 1부터 `n`까지 모든 정수의 합을 출력하세요.

````python
n = 20
# n = int(input())

total = 0
# 1 + 2 + 3 + ... + n
for num in range(n+1):
    # print(num)
    total += num
print(total)
````

````python
n = 5

total = 0
num = 1

while num <= n:
    total += num
    print(num)

    num += 1

print(total)
````

````python
n = 5

total = 0
num = 1

while True:
    total += num
    print(num)

    if num == n:
        break
    num += 1
    

print(total)
````

### 문제 2

지정된 비밀번호(예: `"python123"`)가 입력될 때까지 계속해서 입력을 받는 프로그램을 작성하세요. 틀릴 경우 "다시 시도하세요"를 출력합니다.

````python
valid_pw = 'python123'

while True:
    input_pw = input()

    if input_pw == valid_pw:
        print('성공')
        break
    else:
        print('다시 시도하세요')


# 3번의 시도를 다 실패하면 멈추기.
````

### 문제 3

1부터 30까지의 숫자 중에서 3의 배수를 제외한 나머지 숫자들만 출력하세요.

````python
# 1, 2, 4, 5, 7, 8, 
for num in range(1, 31):
    # if 3의 배수 : 출력하지마
    # if 3의 배수가 아니야 : 출력해.
    if num % 3:
        print(num)
````

````python
num = 1
while num <= 30:
    if num % 3:
        print(num)
    num += 1
````

````python
# 1, 2, 4, 5, 7, 8, 
for num in range(1, 31):
    if num % 3 == 0:
        continue
    print(num)
````

### 문제 4
`names = ['jun', 'ken', 'alex', 'sophia', 'neo']` 일 때,

출석번호를 포함하여 '1번 jun', ..., '5번 neo'의 형태로 출석을 부르시오.

````python
names = ['jun', 'ken', 'alex', 'sophia', 'neo']

for index in range(len(names)):
    name = names[index]
    index += 1
    print(index , name)
    print(f'{index}번 {name}')
````

### 문제 5

```
names = ['jun', 'alex', 'ken']
ages = [18, 28, 37]
``` 
일 때,
```
jun 18
alex 28
ken 37
```
을 출력하시오.

````python
names = ['jun', 'alex', 'ken']
ages = [18, 28, 37]

for index in range(len(names)):
    name = names[index]
    age = ages[index]

    print(name, age)
````

### 문제 6

사용자로부터 숫자를 계속 입력받다가, `0`을 입력하면 지금까지 입력된 모든 숫자의 합계를 출력하고 종료하는 프로그램을 작성하세요.

````python
# 1. 숫자를 계속 입력받는다.
# 2. 그것들을 더한다.
# 3. 0이 나오면 멈춘다.

total = 0

while True:
    input_number = int(input())

    # print(input_number)

    if input_number == 0:
        print(total)
        break

    total += input_number
````

````python
# 1. 숫자를 계속 입력받는다.
# 2. 그것들을 더한다.
# 3. 0이 나오면 멈춘다.

total = 0
input_number = int(input())

while input_number != 0:
    total += input_number

    input_number = int(input())

print(total)
````

### 문제 7

1부터 20까지의 숫자 중 홀수는 건너뛰고, 짝수들만 더하다가 합계가 40을 넘으면 멈추고 합계를 출력하는 프로그램을 작성하세요.

````python
# 1. 1 ~ 20 숫자가 필요하다.
# 2. 홀수는 건너뛰고 짝수를 더한다. -> 짝수를 더한다.
# 3. 합계가 40이 넘으면 멈추고 출력한다.

total = 0

for num in range(1, 21):
    if num % 2 == 0:
        total += num
        print(total)
        if total >= 40:
            print(total)
            break

````

````python
total  = 0
for num in range(1, 21):
    if num % 2:
        continue

    total += num
    print(total)
    if total >= 40:
        print(total)
        break

````
