## BankSim 사기 거래 탐지 데이터

```python
from pathlib import Path
from urllib.request import urlopen
from zipfile import ZipFile
import shutil

import pandas as pd
from IPython.display import display

DATA_DIR = Path("data")
```

### BankSim 데이터

BankSim은 스페인 은행의 거래 집계 자료를 바탕으로 고객과 상점의 결제를 모사한 합성 데이터다. 한 행은 결제 한 건이며, `fraud`는 정상 거래(0)와 사기 거래(1)를 구분하는 정답이다. 총 594,643건의 거래가 있다.

모델의 평가 결과는 시뮬레이션에서 생성된 거래 패턴에 대한 성능을 나타낸다. 실제 금융 거래에 적용하려면 별도의 검증이 필요하다.

출처: Kaggle, Synthetic data from a financial payment system / Edgar Alonso Lopez-Rojas. 원 연구: Lopez-Rojas, E. A. & Axelsson, S. (2014), BankSim: A Bank Payments Simulator for Fraud Detection Research. 라이선스: CC BY-NC-SA 4.0.

### 항목 설명

| 열 | 의미 |
| --- | --- |
| `step` | 시뮬레이션의 시간 단계, 하루 단위 |
| `customer` | 고객 식별자 |
| `age` | 고객 연령대 코드. `0`=18세 이하, `1`=19~25세, `2`=26~35세, `3`=36~45세, `4`=46~55세, `5`=56~65세, `6`=66세 이상, `U`=미상 |
| `gender` | 고객 성별·구분 코드. `F`=여성, `M`=남성, `E`=기업, `U`=미상 |
| `zipcodeOri` | 고객 측 우편번호 |
| `merchant` | 상점 식별자 |
| `zipMerchant` | 상점 우편번호 |
| `category` | 결제 업종 |
| `amount` | 거래 금액 |
| `fraud` | 사기 여부, 0=정상·1=사기 |

### 데이터 다운로드

거래 CSV가 담긴 ZIP을 `data/banksim.zip`에 저장한다.

```python
BANKSIM_URL = "https://www.kaggle.com/api/v1/datasets/download/ealaxi/banksim1/bs140513_032310.csv"
banksim_zip = DATA_DIR / "banksim.zip"

if not banksim_zip.exists():
    temporary_path = banksim_zip.with_suffix(".download")
    with urlopen(BANKSIM_URL, timeout=60) as response:
        with temporary_path.open("wb") as output:
            shutil.copyfileobj(response, output)
    banksim_zip = temporary_path.replace(banksim_zip)
```

### 거래 데이터 불러오기

```python
with ZipFile(banksim_zip) as archive:
    with archive.open("bs140513_032310.csv") as source:
        fraud_df = pd.read_csv(source)

print("전체 행·열:", fraud_df.shape)
display(fraud_df.head(10))
```
