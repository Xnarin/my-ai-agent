## AI·머신러닝·딥러닝·LLM



인공지능은 인식, 예측, 언어 이해처럼 지능이 필요한 작업을 컴퓨터가 수행하도록 만드는 기술이다.
머신러닝은 인공지능을 구현하는 방법 중 하나이며, 딥러닝은 머신러닝에 속한다.
LLM은 딥러닝을 바탕으로 언어를 다루는 모델이다.


### 머신러닝은 무엇인가

머신러닝은 데이터로부터 패턴이나 행동 전략을 학습해 예측·분석·판단·생성 등의 작업을 수행하는 기술이다.
여기서 모델은 입력을 받아 예측 등의 결과를 내는 계산 구조를 말한다.

예를 들어 스팸 메일을 구분하는 규칙을 직접 만든다면 “제목에 무료라는 단어가 있으면 스팸”이라고 정할 수 있다.
하지만 정상 메일에도 그 단어가 들어갈 수 있고, 스팸 메일은 다른 표현을 사용할 수 있다.

모델은 여러 입력 정보와 정답 사이의 관계를 학습하고, 처음 보는 메일이 스팸인지 예측한다.
예측 대상의 정답 값을 레이블, 예측에 사용하는 입력 정보를 특성이라고 한다.

머신러닝은 학습에 어떤 정보를 사용하는지에 따라 여러 방식으로 구분할 수 있다.
대표적인 학습 방식은 다음과 같다.

| 학습 방식 | 학습에 사용하는 정보 | 예시 |
| --- | --- | --- |
| 지도학습 | 입력과 정답 | 집값 예측, 스팸 여부 예측 |
| 비지도학습 | 정답 없이 주어진 데이터 | 구매 행동이 비슷한 고객끼리 묶기 |
| 준지도학습 | 일부 데이터의 정답과 정답 없는 데이터 | 정답이 표시된 일부 사진과 나머지 사진을 함께 활용 |
| 강화학습 | 환경에서 한 행동과 그에 따른 보상 | 게임에서 장기적으로 높은 보상을 얻는 행동 학습 |


K-means는 정답 그룹 없이 비슷한 데이터를 묶는 비지도학습 방법이다.
그림은 두 특성을 가진 설명용 데이터를 그룹 수를 세 개로 정해 묶은 결과다. 각 점은 데이터 한 개이며, 학습 후의 색은 나뉜 그룹을 나타낸다.




### 학습과 추론

학습은 데이터에 맞춰 모델 내부의 값을 조정하는 과정이다.
모델 내부에서 조정되는 값을 파라미터라고 한다.
지도학습에서는 예측과 정답의 차이를 줄이는 방향으로 파라미터를 조정한다.

추론은 학습된 모델에 새로운 입력을 넣어 결과를 얻는 과정이다.
스팸 예시로 모델을 만드는 것은 학습이고, 새로 받은 메일을 분류하는 것은 추론이다.

### 머신러닝은 언제 사용하는가

머신러닝은 규칙을 일일이 작성하기 어렵고, 문제와 관련된 데이터를 확보할 수 있을 때 유용하다.
특히 반복되는 판단이나 예측을 데이터에 근거해 수행할 때 사용한다.

예를 들어 주택 가격에는 면적, 위치, 건축 연도 등 여러 조건이 함께 영향을 준다.
과거 사례에서 이런 조건과 가격의 관계를 학습하는 모델을 사용할 수 있다.

평소와 크게 다른 거래나 설비 상태를 찾아내는 이상탐지에도 머신러닝을 활용한다.


## 딥러닝과 LLM

### 딥러닝은 무엇인가

딥러닝은 신경망을 사용하는 머신러닝 방법이다.
신경망은 입력된 숫자에 계산을 적용해 예측 등의 결과를 만드는 모델이다.
딥러닝에서는 이러한 계산을 여러 층으로 쌓아, 한 층의 결과가 다음 층의 입력으로 이어지도록 한다.

사진에서 고양이를 구분하려면 귀의 모양, 얼굴의 윤곽, 털의 무늬 같은 정보를 활용할 수 있다.
전통적인 머신러닝으로 사진을 분류할 때는 보통 사람이 이런 특징을 수치로 추출하는 방법을 설계하고, 모델은 추출된 특징을 입력받아 분류 기준을 학습한다.
사진마다 자세나 배경이 달라서, 구분에 필요한 특징을 사람이 모두 정하기는 어렵다.

딥러닝에서는 각 픽셀의 밝기나 색을 숫자로 입력하고, 신경망이 고양이를 구분하는 데 유용한 특징과 분류 기준을 함께 학습한다.
앞쪽 층에서는 선이나 경계에 반응하고, 뒤쪽 층에서는 이를 조합한 복잡한 형태에 반응하는 특징이 나타날 수 있다.

각 층의 계산에는 입력에 곱해지는 가중치라는 값이 사용된다.
학습 과정에서는 예측과 정답의 차이가 줄어들도록 가중치 등 모델 내부의 값을 조정한다.





### 딥러닝은 언제 사용하는가

딥러닝은 이미지, 텍스트, 음성 등 다양한 데이터를 다루는 데 사용된다.
사람이 구분이나 예측에 필요한 특징을 하나씩 정하기 어려울 때, 데이터에서 복잡한 패턴을 학습하는 데 유용하다.

| 데이터 | 활용 예 | 학습하는 정보의 예 |
| --- | --- | --- |
| 이미지 | 제품 사진에서 불량 찾기 | 불량을 구분하는 모양이나 질감 |
| 텍스트 | 문장 번역, 글 요약·생성 | 단어와 문장 사이의 관계, 문맥에 맞는 표현 |
| 음성 | 말소리를 글자로 바꾸기 | 소리의 변화와 문자 사이의 관계 |

이런 데이터는 각각 숫자로 표현해 신경망에 입력한다.

딥러닝 모델을 처음부터 학습하려면 많은 데이터와 계산 자원이 필요할 수 있다.
이미 학습된 모델을 가져와 목적에 맞게 추가로 학습하면 필요한 데이터와 학습 비용을 줄일 수 있다.
이처럼 먼저 학습한 지식을 다른 작업에 활용하는 것을 전이학습이라고 한다.

딥러닝을 사용할지는 데이터의 특성과 필요한 예측 성능, 학습과 실행에 드는 비용을 고려해 결정한다.
예를 들어 매출이나 고객 정보 같은 표 데이터에서는 선형 모델이나 트리 기반 모델로도 충분한 성능을 얻을 수 있다.


### LLM은 무엇인가

LLM은 대규모 언어 모델로, 많은 데이터와 큰 규모의 신경망으로 언어의 패턴을 학습한 딥러닝 모델이다.
텍스트는 토큰이라는 처리 단위로 나뉘며, 토큰은 단어나 단어의 일부 등이 될 수 있다.
텍스트를 생성하는 대표적인 LLM은 앞선 토큰들로 다음 토큰을 예측하도록 파라미터를 조정하며 언어의 패턴을 학습한다.
추론할 때는 입력된 문맥을 바탕으로 다음 토큰을 선택하고, 선택한 토큰을 문맥에 추가하는 과정을 반복해 문장을 만든다.

대표적인 LLM은 Transformer라는 신경망 구조를 사용한다.
Transformer는 문맥 속 입력 토큰 사이의 관계를 반영해 정보를 처리하는 구조다.
