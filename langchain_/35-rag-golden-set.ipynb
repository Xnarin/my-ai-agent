### RAG Golden Set 작성 가이드

Golden Set은 RAG 평가의 기준이 되는 질문·기준 답변·원문 근거 모음이다. 평가할 사용자 질문과 원문 범위를 정하고, 문항을 작성한 뒤 원문과 대조하여 확정한다.

**평가 범위 정하기 → 원문·근거 선택 → 질문·기준 답변 작성 → 검수 → 버전 관리·저장**

답변 가능한 문항의 근거는 검색 대상 문서에 포함되어 있어야 하며, 문서 버전도 일치해야 한다. 어떤 사용자의 질문을 다룰지, 어떤 문서와 버전을 기준으로 답할지도 함께 정한다.

### 문항마다 작성할 내용

먼저 평가 대상 원문을 준비하고 문서의 제목, 본문, 출처, 버전을 기록한다. 다음 표는 각 평가 문항에 작성할 수 있는 항목의 예시다. 작성 항목은 고정되어 있지 않으며, 평가 목적과 사용하는 도구에 맞게 구성하고 필요할 때 항목을 추가한다.

| 항목 | 작성할 내용 | 필요한 이유·비고 |
|---|---|---|
| 질문 | RAG에 입력할 사용자 질문 | |
| 기준 답변 | 답변에 반드시 포함되어야 할 핵심 내용과 조건 | 평가할 RAG의 생성 결과가 아니라, 원문 검수로 확정한 비교 기준 |
| 관련 문서 | 답변에 필요한 문서를 확인할 수 있는 파일명·URL·문서 ID | 검색기가 필요한 문서를 찾아왔는지 확인 |
| 근거 구절 | 기준 답변을 뒷받침하는 원문 내용을 그대로 인용 | 기준 답변을 검수하고, 검색된 청크에 실제 답변 근거가 포함됐는지 확인 |
| 근거 위치 | 인용한 내용이 있는 페이지·절·조항 등 | 원문에서 근거를 다시 찾고, 청크 구성이 바뀌어도 같은 근거를 추적 |
| 답변 가능 여부 | 평가 대상 원문 전체로 질문에 답할 수 있는지 여부 | 검색 실패와 원문에 답이 없는 경우를 구분하고, 답변 불가 안내가 적절한지 평가 |
| 질문 유형 | 사실 확인, 조건 비교, 답변 불가 등 | 문항이 특정 유형에 편중되지 않게 구성하고, 유형별 취약점을 확인 |

기준 답변은 출력 문장을 그대로 지정하는 것이 아니라, 답변에 포함되어야 할 내용을 정하는 것이다. 표현이 달라도 핵심 내용과 원문의 조건이 유지되면 된다.

여러 문서가 필요하면 관련 문서와 근거를 모두 작성한다. 문항을 구분할 ID와 평가셋 버전을 함께 관리하고, 관련성 점수는 평가 목적에 따라 추가한다.

평가 결과를 자료 형태나 질문의 특성에 따라 나누어 확인하려면 다음 항목을 선택적으로 추가한다.

| 항목 | 작성할 내용 | 필요한 이유·비고 |
|---|---|---|
| 근거 형태 | 본문 텍스트, 표, 이미지·도표, 혼합 등 | 어떤 자료 형태에서 검색이나 내용 이해에 어려움을 겪는지 확인 |
| 난이도 | 쉬움·보통·어려움 등 | 난이도별 성능을 비교하고, 쉬운 문항에 편중되지 않게 구성 |
| 복수 출처 필요 여부 | 한 문서로 답할 수 있는지, 여러 문서의 근거를 함께 봐야 하는지 | 여러 출처의 근거를 검색하고 종합하는 능력을 평가 |

난이도는 필요한 정보의 수, 조건·예외 해석, 여러 근거의 종합 여부 등을 기준으로 미리 구분해 둔다. 복수 출처 필요 여부는 같은 문서의 여러 청크를 사용하는 경우와 구분한다.

원문 문서와 근거 위치를 연결해 두면 청크 크기를 바꿔도 같은 기준으로 비교할 수 있다. 청크 ID를 정답으로 지정하면 필요한 근거를 검색했는지 더 세밀하게 평가할 수 있다. 다만 재청킹할 때마다 정답 청크 ID가 달라지므로 평가셋도 갱신해야 한다.

필드 이름과 저장 형식은 사용하는 평가 도구에 맞춘다.

### 질문과 기준 답변 작성하기

답변 가능한 문항은 먼저 확인할 사실과 근거를 고른 뒤, 그 내용을 묻는 질문과 기준 답변을 작성한다. 실제 사용자 질문이나 기존 실패 사례에서 시작한 경우에도 원문 근거를 확인한다.

- 질문만 읽어도 무엇을 묻는지 알 수 있게 작성한다.
- 질문에 정답을 노출하지 않고, 실제 사용자가 사용할 만한 표현을 쓴다.
- 기준 답변에는 필요한 조건·수치·예외를 포함하고 원문 밖의 내용을 추가하지 않는다.
- 사실 확인, 조건 비교, 여러 문서를 함께 봐야 하는 질문, 유사한 용어를 구별하는 질문 등을 구성한다.
- 답변 불가 질문과 실제 사용자 질문, 기존 RAG가 잘못 답한 질문도 포함한다.

같은 사실을 표현만 바꾸어 반복하거나 특정 문서에 편중되지 않도록 한다. 질문 유형과 비중은 실제 사용 상황과 확인하려는 실패 유형에 맞춘다.

### 문항 작성 예시

다음은 가상의 도서관 안내 문서다. 이 예시에서는 아래 내용이 검색 대상 문서의 전부다.

- 문서 제목: 도서 대출 안내
- 문서 ID: `library-guide`
- 문서 버전: `v1`
- 근거 위치: 대출 기간과 연장 절

> 일반 회원은 도서를 최대 3권까지 14일 동안 대출할 수 있다. 예약자가 없는 도서는 1회에 한해 7일 연장할 수 있다.

이 원문으로 작성한 평가셋 `v1`의 문항 예시는 다음과 같다.

| 항목 | 답변 가능한 문항 | 답변 불가 문항 |
|---|---|---|
| 문항 ID | `loan-extension` | `overdue-fee` |
| 질문 | 일반 회원이 빌린 도서는 어떤 조건에서 얼마나 연장할 수 있나요? / 며칠 더 빌릴 수 있어요? | 도서 반납이 늦어지면 연체료는 하루에 얼마인가요? / 반납 늦으면 얼마 내요? |
| 기준 답변 | 예약자가 없는 도서에 한해 1회, 7일 연장할 수 있다. | 제공된 문서에는 연체료 정보가 없어 하루 금액을 확인할 수 없다고 안내한다. |
| 관련 문서 | `["library-guide"]` | `[]` |
| 근거 구절 | 예약자가 없는 도서는 1회에 한해 7일 연장할 수 있다. | `[]` |
| 근거 위치 | 대출 기간과 연장 절 | 해당 없음 |
| 답변 가능 여부 | 가능 | 불가 |
| 질문 유형 | 사실 확인 | 답변 불가 |

첫 문항은 예약자 유무, 연장 횟수, 연장 기간을 모두 포함해야 한다. 두 번째 문항은 연체료 금액을 추측하거나 무료라고 단정하지 않고 정보가 없음을 안내해야 한다.

### LLM으로 초안 만들기

답변 가능한 문항의 초안을 만들 때는 LLM에 사용할 원문과 문서 ID, 질문 유형, 기존 질문, 원하는 응답 항목을 함께 제공한다. 프롬프트에는 다음 작성 기준을 명시한다.

- 제공된 원문으로 답할 수 있는 질문을 작성한다.
- 질문은 단독으로 이해할 수 있어야 하며, 기존 질문과 같은 사실을 반복하지 않는다.
- 기준 답변에 원문에 없는 조건이나 수치를 추가하지 않는다.
- 기준 답변의 모든 핵심 내용을 뒷받침하는 원문 구절을 그대로 인용한다.
- 질문·기준 답변·근거를 구분해서 반환한다.

출처는 준비한 문서와 연결한다. 응답 형식을 고정해도 내용의 정확성이 보장되지는 않는다. 생성 결과는 초안으로 취급하고 원문 검수를 거친 문항만 평가에 사용한다.

### 원문과 대조하여 검수하기

| 확인 방식 | 확인할 내용 |
|---|---|
| 자동 검사 | 문항 유형별 필수 항목 누락, ID·질문 중복, 존재하지 않는 출처, 근거 구절의 원문 포함 여부 |
| 사람 검수 | 질문의 자연스러움, 답변의 정확성·완전성, 조건·예외 누락, 출처 누락, 의미상 중복 |

근거 구절이 원문에 있는지와 그 근거가 기준 답변을 뒷받침하는지를 모두 확인한다. 답변의 수치·조건·적용 범위가 원문과 일치하고 필요한 예외가 빠지지 않았는지 검토한다.

관련 문서를 찾았더라도 검색된 청크에 답변에 필요한 근거가 없을 수 있다. 문서의 일치 여부와 실제 근거의 포함 여부를 구분한다.

오류가 있는 문항은 수정하거나 제외하고, 수정한 문항은 다시 원문과 대조한다.

### 답변 불가 문항 구성하기

답변 가능 여부는 일부 청크가 아니라 평가 대상 원문 문서 모음 전체를 기준으로 판단한다. 검색기가 근거를 찾지 못한 경우와 원문에 답이 없는 경우를 구분한다.

답변 불가 문항에는 “제공된 문서로는 확인할 수 없습니다.”처럼 기대하는 응답 방식을 기록한다. 정답 근거가 없는 문항의 관련 문서와 근거 목록은 비워 두고, 없는 정답이나 근거를 만들어 넣지 않는다.

관련 문서 목록이 비어 있는 문항은 Recall@k의 분모가 0이 되므로 해당 지표의 평균에서 제외하고 제외 수를 기록한다. 응답은 별도로 평가하여 근거 부족을 안내하는지, 없는 사실을 만들어 답하지 않는지 확인한다.

### 평가셋 분리와 버전 관리

- 개발셋은 청크 크기, 검색 방식, 프롬프트를 바꾸며 반복 비교할 때 사용한다.
- 테스트셋은 개선 과정에 사용하지 않고, 설정을 확정한 뒤 최종 성능을 확인할 때 사용한다.

동일한 문항의 결과만 보며 개선하면 그 문항에 맞춰질 수 있으므로 최종 확인용 문항을 구분한다.

검수한 문항은 JSON·JSONL 등으로 저장한다. 원문과 평가셋의 버전을 함께 기록하고, 작성 당시의 원문을 확인할 수 있도록 보관한다. 질문·정답·근거가 바뀌면 새 버전으로 관리한다.

### RAG 평가에 연결하기

RAG에는 평가 질문을 입력하고, 실제 검색 결과와 생성 답변을 기록한다. 준비한 관련 문서·근거·기준 답변과 비교하여 검색과 답변 품질을 평가한다.

| 미리 준비하는 기준 | 실행 후 기록하는 결과 |
|---|---|
| 관련 문서 ID와 원문 근거 | 실제 검색한 문서·청크, 출처 ID, 검색 순위 |
| 기준 답변 | RAG가 실제로 생성한 답변 |

기준 근거를 실제 검색 결과 대신 넣거나 질문·정답 파일을 검색 대상 원문에 추가하지 않는다. 비교할 때는 같은 버전의 Golden Set을 사용하고, 평가 결과에도 해당 버전을 기록한다.
